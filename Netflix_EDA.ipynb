```python
df['date_added'] = pd.to_datetime(
    df['date_added'],
    errors='coerce'
)
```

```python
df['country'] = df['country'].fillna('Unknown')
df['director'] = df['director'].fillna('Unknown')
df['cast'] = df['cast'].fillna('Unknown')
```

```python
df.head()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="type"
)

plt.title("Distribution of Movies and TV Shows")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")

plt.tight_layout()
plt.show()
```

```python
release_year_counts = (
    df["release_year"]
    .value_counts()
    .sort_index()
)

plt.figure(figsize=(12, 5))

plt.plot(
    release_year_counts.index,
    release_year_counts.values,
    marker="o"
)

plt.title("Number of Titles by Release Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")

plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
recent = df[df["release_year"] >= 2000]

year_counts = (
    recent["release_year"]
    .value_counts()
    .sort_index()
)

plt.figure(figsize=(12, 5))

plt.plot(
    year_counts.index,
    year_counts.values,
    marker="o"
)

plt.title("Netflix Titles Released Since 2000")
plt.xlabel("Year")
plt.ylabel("Number of Titles")

plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
year_type = pd.crosstab(
    df["release_year"],
    df["type"]
)

year_type = year_type.loc[
    year_type.index >= 2000
]

year_type.plot(
    figsize=(12, 6),
    marker="o"
)

plt.title("Movies and TV Shows by Release Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")

plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
